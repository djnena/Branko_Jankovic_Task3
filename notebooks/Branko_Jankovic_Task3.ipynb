# Task - Predikcija kategorije proizvoda na osnovu naslova


## Ucitavanje i analiza podataka


```python
import pandas as pd
import matplotlib.pyplot as plt

# load dataset from GitHub
url = "https://raw.githubusercontent.com/djnena/Branko_Jankovic_Task3/main/data/products.csv"
df = pd.read_csv(url)

# Print shape (number of rows and columns)
print("Dataset shape (rows, columns):", df.shape)

# Show first 5 rows
print("\nFirst 5 rows:")
display(df.head())

# Show column data types and non-null counts
print("\nDataset info:")
df.info()
```

## Uoceni problemi i nedoslednosti
### 1. Nedostajuce vrednosti

```python
# Broj nedostajućih vrednosti po kolonama
print(df.isnull().sum())
```

### 2. Razmaci u nazivima kolona

Neke kolone imaju vodece ili zavrsne razmake, na primer:

" Category Label"
" Listing Date"

To moze izazvati greske pri pisanju koda.

```python
df.columns = df.columns.str.strip()
```

### 3. Neodgovarajuci tip podataka

Kolona Listing Date je ucitana kao tekst (object), iako predstavlja datum.

Potrebno ju je konvertovati u datetime format:

```python
df["Listing Date"] = pd.to_datetime(df["Listing Date"])
```

### 4. Veliki broj jedinstvenih vrednosti

```python
# Broj jedinstvenih vrednosti po kolonama
print(df.nunique())
```

### 5. Numericki uvid u raspodelu sentimenta

```python
# Count occurrences of each sentiment label
category_counts = df['Category Label'].value_counts()

# Print counts
print("Category distribution (counts):")
print(category_counts)
```

## Za pripremu i ciscenje podataka pre modeliranja mozes primeniti sledece korake:

```python
#brisanje nedostajucih vrednosti
df=df.dropna()

# Zamena vrednosti
df["Category Label"] = df["Category Label"].replace("fridge", "Fridges").replace("CPU","CPUs").replace("Mobile Phone","Mobile Phones")
```

Graficki prikaz:

```python
# Grafički prikaz
plt.figure(figsize=(10,6))
category_counts = df['Category Label'].value_counts()
category_counts.plot(kind="bar")

plt.title("Raspodela proizvoda po kategorijama")
plt.xlabel("Kategorija")
plt.ylabel("Broj proizvoda")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```
